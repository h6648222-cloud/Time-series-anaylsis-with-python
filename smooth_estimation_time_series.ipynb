```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#data
at = [
320,330,325,335,340,345,342,348,350,355,
360,362,365,368,370,372,375,378,380,385,
388,390,392,395,398,400,402,405,408,410,
412,415,418,420,423,425,428,430,432,435,
438,440,443,445,448,450,452,455,458,460,
462,465,468,470,473,475,478,480,482,485,
488,490,493,495,498,500,502,505,508,510,
512,515,518,520,523,525,528,530,532,535,
538,540,543,545,548,550,552,555,558,560,
562,565,568,570,573,575,578,580,585,590
]
```

```python
#forcast alpha at 0.3
alpha = 0.3
#make vector that can store the  forcating data
S1 = [0]*len(at)
S1[0] = at[0] # Initialize first forecast with first actual value
for i in range(1,len(at)):
  S1[i] = alpha*at[i]+(1-alpha)*S1[i-1]
```

intialize alpha
an empty vector formation
for loop equal to the length of data
S1[i]= formula of smooth estimation

```python
# forcast alpha at 0.7
alpha = 0.7
S2 = [0]*len(at)
S2[0] = at[0] # Initialize first forecast with first actual value
for j in range(1,len(at)):
  S2[j]=alpha*at[j]+(1-alpha)*S2[j-1]
```

```python
# output with pandas
df =pd.DataFrame({
    'Month':range(1,len(at)+1),
    'Actual':at,
    'Forcast_S1':S1,
    'Forcast_S2':S2
})
df
```

```python
Month=range(1,len(at)+1)
actual = at
plt.plot(Month,actual,label='Actual',marker='o')
# symbol used to show each individual data point marker
plt.plot(Month,S1,label='forcast_alpha_0.3',marker='x')
plt.xlabel("Month")
plt.ylabel("Value")
plt.title("Exponential Smoothing Forecast vs Actual")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(Month,actual,label='Actual',marker='^',color='red')
plt.plot(Month,S2,label='forcast_of_alpha_0.7',marker='o',color='limegreen')
plt.xlabel='Months'
plt.ylabel='Value'
plt.legend()
plt.title("smooth estimation alpha 0.7")
plt.grid(True)
plt.show()
```
